Experiment 9: File System Feature Simulation (FAT vs NTFS vs EXT)

```python
# Experiment 9: File System Feature Simulation

# Define file system features
file_systems = {
    "FAT": {
        "File": "document.txt",
        "Permissions": "No",
        "Journaling": "No"
    },
    "NTFS": {
        "File": "document.txt",
        "Permissions": "Yes",
        "Journaling": "Yes"
    },
    "EXT": {
        "File": "document.txt",
        "Permissions": "Yes",
        "Journaling": "Yes"
    }
}

print("FILE SYSTEM FEATURE SIMULATION")
print("-" * 60)

print("{:<10} {:<15} {:<15} {:<15}".format(
    "File System", "File", "Permissions", "Journaling"))
print("-" * 60)

for fs, details in file_systems.items():
    print("{:<10} {:<15} {:<15} {:<15}".format(
        fs,
        details["File"],
        details["Permissions"],
        details["Journaling"]
    ))

print("\nComparison:")
print("FAT  : No file permissions and no journaling.")
print("NTFS : Supports file permissions and journaling.")
print("EXT  : Supports file permissions and journaling.")
```
