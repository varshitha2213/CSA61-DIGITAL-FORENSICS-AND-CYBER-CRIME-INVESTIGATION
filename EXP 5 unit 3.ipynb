Experiment 5: Imaging vs Duplication

```python
import os
import shutil

# -------------------------------
# Create Sample Storage Device
# -------------------------------
source_folder = "Storage_Device"
os.makedirs(source_folder, exist_ok=True)

# Active (visible) files
with open(os.path.join(source_folder, "document.txt"), "w") as f:
    f.write("Project Report")

with open(os.path.join(source_folder, "photo.jpg"), "w") as f:
    f.write("Image Data")

# Simulated deleted data
with open(os.path.join(source_folder, "deleted_data.bin"), "w") as f:
    f.write("Deleted File Data")

# -------------------------------
# Forensic Imaging
# -------------------------------
image_folder = "Forensic_Image"
os.makedirs(image_folder, exist_ok=True)

for file in os.listdir(source_folder):
    shutil.copy(os.path.join(source_folder, file), image_folder)

print("FORENSIC IMAGING")
print("----------------------------")
print("Captured all sectors (including deleted data).")
print("Files Captured:")
print(os.listdir(image_folder))

# -------------------------------
# Ordinary Duplication
# -------------------------------
duplicate_folder = "Ordinary_Duplication"
os.makedirs(duplicate_folder, exist_ok=True)

for file in os.listdir(source_folder):
    if file != "deleted_data.bin":     # Skip deleted data
        shutil.copy(os.path.join(source_folder, file), duplicate_folder)

print("\nORDINARY DUPLICATION")
print("----------------------------")
print("Copied only active (visible) files.")
print("Files Copied:")
print(os.listdir(duplicate_folder))
```
