Experiment 7: Hashing & the Avalanche Effect.

```python
import hashlib

# Original and modified text
text1 = "hello"
text2 = "Hello"

# Function to generate hashes
def generate_hashes(text):
    md5 = hashlib.md5(text.encode()).hexdigest()
    sha1 = hashlib.sha1(text.encode()).hexdigest()
    sha256 = hashlib.sha256(text.encode()).hexdigest()

    print("Input :", text)
    print("MD5    :", md5)
    print("SHA-1  :", sha1)
    print("SHA-256:", sha256)
    print("-" * 60)

# Generate hashes
print("HASHING & AVALANCHE EFFECT")
print("=" * 60)

generate_hashes(text1)
generate_hashes(text2)

# Compare the hashes
print("Result:")
print("Changing one character ('hello' → 'Hello') produces completely different hash values.")
print("This demonstrates the Avalanche Effect.")
```
