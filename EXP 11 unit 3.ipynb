Experiment 11: End-to-End Forensic Workflow

```python
import hashlib
import shutil

# -------------------------------
# Step 1: Acquire Evidence
# -------------------------------
original_file = "evidence.txt"

with open(original_file, "w") as file:
    file.write("This is the original digital evidence.")

acquired_file = "acquired_evidence.txt"
shutil.copy(original_file, acquired_file)

print("Step 1: Acquire")
print("Evidence acquired successfully.\n")

# -------------------------------
# Function to Calculate SHA-256
# -------------------------------
def calculate_hash(filename):
    sha = hashlib.sha256()

    with open(filename, "rb") as file:
        while True:
            data = file.read(4096)
            if not data:
                break
            sha.update(data)

    return sha.hexdigest()

# -------------------------------
# Step 2: Hash
# -------------------------------
original_hash = calculate_hash(acquired_file)

print("Step 2: Hash")
print("SHA-256:", original_hash)

# -------------------------------
# Step 3: Verify
# -------------------------------
verify_hash = calculate_hash(acquired_file)

print("\nStep 3: Verify")

if original_hash == verify_hash:
    print("Integrity Verified.")
else:
    print("Integrity Verification Failed.")

# -------------------------------
# Step 4: Analyze
# -------------------------------
print("\nStep 4: Analyze")
print("Evidence Size:", len(open(acquired_file).read()), "bytes")
print("No suspicious modification detected.")

# -------------------------------
# Step 5: Report
# -------------------------------
print("\nStep 5: Report")
print("Investigation completed successfully.")

# -------------------------------
# Tampering with Evidence
# -------------------------------
print("\nTampering with Evidence...")

with open(acquired_file, "a") as file:
    file.write("\nEvidence Modified")

tampered_hash = calculate_hash(acquired_file)

print("\nVerifying After Tampering...")

if original_hash == tampered_hash:
    print("Integrity Verified.")
else:
    print("Integrity Verification Failed!")
    print("Evidence has been altered.")
```
