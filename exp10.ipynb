```python
headers = {
    "From": "bank@secure.com",
    "Return-Path": "hacker@fake.com",
    "Received": "mail.fake.com"
}

print("Email Header Analysis\n")

print("From:", headers["From"])
print("Return-Path:", headers["Return-Path"])

if headers["From"] != headers["Return-Path"]:
    print("\nPossible Email Spoofing Detected")
else:
    print("\nNo Spoofing Detected")
```
