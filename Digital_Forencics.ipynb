```python
import os
import time

filename = "sample.txt"

# Create file if it doesn't exist
with open(filename, "a") as f:
    f.write("Original Data\n")

print("Before Modification")
print("Created :", time.ctime(os.path.getctime(filename)))
print("Modified:", time.ctime(os.path.getmtime(filename)))
print("Accessed:", time.ctime(os.path.getatime(filename)))

input("\nPress Enter to modify the file...")

# Modify file
with open(filename, "a") as f:
    f.write("Evidence Modified\n")

print("\nAfter Modification")
print("Created :", time.ctime(os.path.getctime(filename)))
print("Modified:", time.ctime(os.path.getmtime(filename)))
print("Accessed:", time.ctime(os.path.getatime(filename)))
```

```python
import hashlib

filename = input("Enter file name: ")

def calculate_hash(file):
    sha = hashlib.sha256()

    with open(file, "rb") as f:
        while True:
            data = f.read(4096)
            if not data:
                break
            sha.update(data)

    return sha.hexdigest()

original_hash = calculate_hash(filename)

print("\nOriginal Hash:")
print(original_hash)

new_hash = calculate_hash(filename)

print("\nVerification Hash:")
print(new_hash)

if original_hash == new_hash:
    print("\nIntegrity Verified")
else:
    print("\nFile Tampered")
```

```python
sender = input("Enter sender email: ")
subject = input("Enter subject: ")
body = input("Enter email body: ")

score = 0

keywords = ["urgent", "verify", "password", "bank", "click", "login"]

if not sender.endswith("@gmail.com") and not sender.endswith("@outlook.com"):
    score += 1

for word in keywords:
    if word.lower() in subject.lower():
        score += 1

for word in keywords:
    if word.lower() in body.lower():
        score += 1

print("\nPhishing Score:", score)

if score >= 3:
    print("Potential Phishing Email")
else:
    print("Likely Safe Email")
```
