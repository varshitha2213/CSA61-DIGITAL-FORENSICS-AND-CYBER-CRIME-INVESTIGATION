1. Demonstrate Locard's Exchange Principle using file timestamps

```python
import os
import time

filename = "sample.txt"

# Create file if it doesn't exist
with open(filename, "a") as f:
    f.write("Original Data\n")

print("Before Modification")
print("Created :", time.ctime(os.path.getctime(filename)))
print("Modified:", time.ctime(os.path.getmtime(filename)))
print("Accessed:", time.ctime(os.path.getatime(filename)))

input("\nPress Enter to modify the file...")

# Modify file
with open(filename, "a") as f:
    f.write("Evidence Modified\n")

print("\nAfter Modification")
print("Created :", time.ctime(os.path.getctime(filename)))
print("Modified:", time.ctime(os.path.getmtime(filename)))
print("Accessed:", time.ctime(os.path.getatime(filename)))
```

2. Verify the Chain of Custody of Digital Evidence using File Hashing

```python
import hashlib

filename = input("Enter file name: ")

def calculate_hash(file):
    sha = hashlib.sha256()

    with open(file, "rb") as f:
        while True:
            data = f.read(4096)
            if not data:
                break
            sha.update(data)

    return sha.hexdigest()

original_hash = calculate_hash(filename)

print("\nOriginal Hash:")
print(original_hash)

new_hash = calculate_hash(filename)

print("\nVerification Hash:")
print(new_hash)

if original_hash == new_hash:
    print("\nIntegrity Verified")
else:
    print("\nFile Tampered")
```

3. Detect Phishing Emails Based on Suspicious Indicators

```python
sender = input("Enter sender email: ")
subject = input("Enter subject: ")
body = input("Enter email body: ")

score = 0

keywords = ["urgent", "verify", "password", "bank", "click", "login"]

if not sender.endswith("@gmail.com") and not sender.endswith("@outlook.com"):
    score += 1

for word in keywords:
    if word.lower() in subject.lower():
        score += 1

for word in keywords:
    if word.lower() in body.lower():
        score += 1

print("\nPhishing Score:", score)

if score >= 3:
    print("Potential Phishing Email")
else:
    print("Likely Safe Email")
```

6. Create a Bit-Stream (Forensic) Copy of a File and Verify Integrity

```python
import hashlib

source = input("Enter source file: ")
destination = "forensic_copy.bin"

with open(source, "rb") as src:
    data = src.read()

with open(destination, "wb") as dest:
    dest.write(data)

print("Bit-stream copy created.")

def file_hash(file):
    sha = hashlib.sha256()

    with open(file, "rb") as f:
        while True:
            chunk = f.read(4096)
            if not chunk:
                break
            sha.update(chunk)

    return sha.hexdigest()

hash1 = file_hash(source)
hash2 = file_hash(destination)

print("\nOriginal Hash :", hash1)
print("Copied Hash   :", hash2)

if hash1 == hash2:
    print("Integrity Verified")
else:
    print("Integrity Failed")
```

7. Identify a File's True Type from its Byte-Level Signature (Magic Number)

```python
from google.colab import files

uploaded = files.upload()
```

```python
magic_numbers = {
    b'\xFF\xD8\xFF': "JPEG Image",
    b'\x89PNG': "PNG Image",
    b'%PDF': "PDF Document",
    b'PK\x03\x04': "ZIP Archive",
    b'GIF87a': "GIF Image",
    b'GIF89a': "GIF Image"
}

filename = input("Enter file name: ")

with open(filename, "rb") as f:
    header = f.read(8)

found = False

for signature, filetype in magic_numbers.items():
    if header.startswith(signature):
        print("File Type:", filetype)
        found = True
        break

if not found:
    print("Unknown File Type")
```
