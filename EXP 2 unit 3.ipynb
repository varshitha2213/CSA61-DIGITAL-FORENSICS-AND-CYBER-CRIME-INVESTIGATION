Experiment 2: Volatile vs Non-Volatile Evidence Capture

```python
import hashlib
import os

# -------------------------------
# Volatile Evidence (Simulated)
# -------------------------------
print("VOLATILE EVIDENCE")
print("----------------------------")

running_processes = [
    "chrome.exe",
    "python.exe",
    "explorer.exe",
    "notepad.exe"
]

print("Running Processes:")
for process in running_processes:
    print(process)

print("\n(Note: Volatile evidence is lost after system shutdown.)")

# -------------------------------
# Non-Volatile Evidence
# -------------------------------
print("\nNON-VOLATILE EVIDENCE")
print("----------------------------")

filename = "evidence.txt"

# Create a sample file
with open(filename, "w") as file:
    file.write("This is non-volatile digital evidence.")

# Function to calculate MD5 hash
def md5_hash(file_name):
    with open(file_name, "rb") as file:
        data = file.read()
        return hashlib.md5(data).hexdigest()

# Read the file twice
hash1 = md5_hash(filename)
hash2 = md5_hash(filename)

print("File Name :", filename)
print("File Size :", os.path.getsize(filename), "bytes")
print("First MD5 :", hash1)
print("Second MD5:", hash2)

# Check integrity
if hash1 == hash2:
    print("\nResult: Non-volatile evidence remains identical when re-read.")
else:
    print("\nResult: File has been modified.")
```
