Experiment 10: Storage Media Analysis (Slack Space Calculation).

```python
# Experiment 10: Storage Media Analysis
# Aim: Calculate Slack Space

# Input values
file_size = int(input("Enter file size (in bytes): "))
cluster_size = int(input("Enter cluster size (in bytes): "))

# Calculate allocated space
allocated_space = ((file_size + cluster_size - 1) // cluster_size) * cluster_size

# Calculate slack space
slack_space = allocated_space - file_size

# Display results
print("\nSTORAGE MEDIA ANALYSIS")
print("-" * 30)
print("File Size       :", file_size, "bytes")
print("Cluster Size    :", cluster_size, "bytes")
print("Allocated Space :", allocated_space, "bytes")
print("Slack Space     :", slack_space, "bytes")
```
