```python
# ==========================================
# Experiment 1.2 - Malware Hash Analysis
# ==========================================

import hashlib
import pandas as pd

print("="*60)
print("      MALWARE HASH ANALYSIS")
print("="*60)

# ------------------------------------------
# Create a Sample File
# ------------------------------------------

filename = "sample_file.txt"

with open(filename, "w") as f:
    f.write("This is a sample file used for malware hash analysis.")

print("\nSample File Created:", filename)

# ------------------------------------------
# Read File
# ------------------------------------------

with open(filename, "rb") as f:
    data = f.read()

# ------------------------------------------
# Generate Hashes
# ------------------------------------------

md5 = hashlib.md5(data).hexdigest()
sha1 = hashlib.sha1(data).hexdigest()
sha256 = hashlib.sha256(data).hexdigest()

print("\nGenerated Hashes")
print("-"*50)
print("MD5    :", md5)
print("SHA1   :", sha1)
print("SHA256 :", sha256)

# ------------------------------------------
# Malware Hash Database
# ------------------------------------------

database = [
    {
        "Family":"Trojan",
        "Algorithm":"MD5",
        "Hash":"11111111111111111111111111111111"
    },
    {
        "Family":"WannaCry",
        "Algorithm":"SHA1",
        "Hash":"aaaaaaaaaaaaaaaaaaaaaaaaaaaaaaaaaaaaaaaa"
    },
    {
        "Family":"Emotet",
        "Algorithm":"SHA256",
        "Hash":"bbbbbbbbbbbbbbbbbbbbbbbbbbbbbbbbbbbbbbbbbbbbbbbbbbbbbbbbbbbbbbbb"
    },
    {
        "Family":"Sample_File",
        "Algorithm":"SHA256",
        "Hash":sha256
    }
]

df = pd.DataFrame(database)

print("\nMalware Database")
print("-"*50)
print(df)

df.to_csv("malware_hash_database.csv", index=False)

# ------------------------------------------
# Detection
# ------------------------------------------

print("\nDetection Result")
print("-"*50)

detected = False

for index,row in df.iterrows():

    if row["Algorithm"]=="MD5" and row["Hash"]==md5:
        print("Known Malware Detected :",row["Family"])
        detected=True

    elif row["Algorithm"]=="SHA1" and row["Hash"]==sha1:
        print("Known Malware Detected :",row["Family"])
        detected=True

    elif row["Algorithm"]=="SHA256" and row["Hash"]==sha256:
        print("Known Malware Detected :",row["Family"])
        detected=True

if not detected:
    print("No Known Malware Detected")

# ------------------------------------------
# Statistics
# ------------------------------------------

print("\nDetection Statistics")
print("-"*50)

total = len(df)

matched = 1 if detected else 0

rate = (matched/total)*100

print("Total Malware Signatures :",total)
print("Matched Signatures       :",matched)
print("Detection Rate           : {:.2f}%".format(rate))

# ------------------------------------------
# Hash Family Count
# ------------------------------------------

print("\nHash Family Clustering")
print("-"*50)

print(df["Family"].value_counts())

print("\nCSV File Saved : malware_hash_database.csv")
```
