```python
import re
import pandas as pd

password_attempts=[
"123456",
"admin",
"password",
"qwerty",
"letmein",
"Secure@123"
]

correct_password="Secure@123"

logs=[]

for i,pwd in enumerate(password_attempts,1):

    if pwd==correct_password:
        logs.append([i,pwd,"Success"])
        break
    else:
        logs.append([i,pwd,"Failed"])

sql_payloads=[
"' OR '1'='1",
"admin' --",
"SELECT * FROM users",
"NormalInput"
]

directory_payloads=[
"../../etc/passwd",
"..\\Windows\\System32",
"index.html"
]

sql_detection=[]

for payload in sql_payloads:

    if re.search(r"('|--|select|or)",payload,re.IGNORECASE):
        sql_detection.append([payload,"Detected"])
    else:
        sql_detection.append([payload,"Safe"])

dir_detection=[]

for payload in directory_payloads:

    if "../" in payload or "..\\" in payload:
        dir_detection.append([payload,"Traversal Detected"])
    else:
        dir_detection.append([payload,"Safe"])

print("Brute Force Log")
print(pd.DataFrame(logs,
columns=["Attempt","Password","Result"]))

print("\nSQL Injection Detection")
print(pd.DataFrame(sql_detection,
columns=["Payload","Status"]))

print("\nDirectory Traversal Detection")
print(pd.DataFrame(dir_detection,
columns=["Payload","Status"]))

print("\nSuggested WAF Rules")
print("1. Block SQL keywords (SELECT, UNION, OR)")
print("2. Block '../' directory traversal")
print("3. Limit login attempts to 5")
print("4. Enable CAPTCHA after multiple failures")
```
