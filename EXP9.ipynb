```python
from collections import defaultdict

login_logs = [
    "192.168.1.10 FAILED",
    "192.168.1.10 FAILED",
    "192.168.1.10 FAILED",
    "192.168.1.10 FAILED",
    "192.168.1.15 SUCCESS",
    "192.168.1.20 FAILED",
    "192.168.1.20 FAILED",
    "192.168.1.20 SUCCESS"
]

failed_attempts = defaultdict(int)

for log in login_logs:
    ip, status = log.split()

    if status == "FAILED":
        failed_attempts[ip] += 1
    else:
        failed_attempts[ip] = 0

print("Brute Force Detection Report")
print("-" * 35)

threshold = 3

for ip, count in failed_attempts.items():
    if count >= threshold:
        print(f"Alert: {ip} made {count} failed login attempts.")
```
