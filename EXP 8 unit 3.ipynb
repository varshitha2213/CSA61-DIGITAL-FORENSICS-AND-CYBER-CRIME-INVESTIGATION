Experiment 8: File Carving (Data Recovery)

```python
# Experiment 8: File Carving (Data Recovery)

# Create a raw byte blob with two simulated JPEG files
raw_data = (
    b"RANDOMDATA"
    + b"\xFF\xD8\xFF" + b"JPEG IMAGE 1 DATA" + b"\xFF\xD9"
    + b"GARBAGE DATA"
    + b"\xFF\xD8\xFF" + b"JPEG IMAGE 2 DATA" + b"\xFF\xD9"
)

# Save the raw data
with open("raw_blob.bin", "wb") as file:
    file.write(raw_data)

# Read the raw data
with open("raw_blob.bin", "rb") as file:
    data = file.read()

start_signature = b"\xFF\xD8\xFF"
end_signature = b"\xFF\xD9"

count = 0
position = 0

while True:
    start = data.find(start_signature, position)
    if start == -1:
        break

    end = data.find(end_signature, start)

    if end == -1:
        break

    end += len(end_signature)

    recovered = data[start:end]

    filename = f"Recovered_Image_{count + 1}.jpg"

    with open(filename, "wb") as img:
        img.write(recovered)

    print(filename, "Recovered Successfully")

    count += 1
    position = end

print("\nTotal Files Recovered:", count)
```
