```python
Experiment 3: Physical, Logical and Sparse Acquisition
```

```python
import os
import shutil

# Create sample evidence folder
source_folder = "Disk"
os.makedirs(source_folder, exist_ok=True)

# Create sample files
files = {
    "document.txt": "Project Report",
    "image.jpg": "Image Data",
    "email.eml": "Email Evidence",
    "deleted.dat": "Deleted File Data (Recovered)"
}

for filename, content in files.items():
    with open(os.path.join(source_folder, filename), "w") as f:
        f.write(content)

# -------------------------------
# Physical Acquisition
# -------------------------------
physical_folder = "Physical_Acquisition"
os.makedirs(physical_folder, exist_ok=True)

for file in os.listdir(source_folder):
    shutil.copy(os.path.join(source_folder, file), physical_folder)

print("PHYSICAL ACQUISITION")
print("Captured all files, including deleted/unused data.")
print(os.listdir(physical_folder))

# -------------------------------
# Logical Acquisition
# -------------------------------
logical_folder = "Logical_Acquisition"
os.makedirs(logical_folder, exist_ok=True)

for file in os.listdir(source_folder):
    if file != "deleted.dat":          # Deleted file excluded
        shutil.copy(os.path.join(source_folder, file), logical_folder)

print("\nLOGICAL ACQUISITION")
print("Captured only visible files.")
print(os.listdir(logical_folder))

# -------------------------------
# Sparse Acquisition
# -------------------------------
sparse_folder = "Sparse_Acquisition"
os.makedirs(sparse_folder, exist_ok=True)

selected_files = ["document.txt", "email.eml"]

for file in selected_files:
    shutil.copy(os.path.join(source_folder, file), sparse_folder)

print("\nSPARSE ACQUISITION")
print("Captured only selected files.")
print(os.listdir(sparse_folder))
```
