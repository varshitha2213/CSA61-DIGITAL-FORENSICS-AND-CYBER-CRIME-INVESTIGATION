Experiment 4: Live vs Dead Acquisition.

```python
import os

# -------------------------------
# Live Acquisition (System ON)
# -------------------------------
print("LIVE ACQUISITION")
print("-------------------------")

# Simulated running processes (volatile data)
running_processes = [
    "chrome.exe",
    "python.exe",
    "explorer.exe",
    "cmd.exe"
]

print("Running Processes:")
for process in running_processes:
    print(process)

# Save a disk snapshot
snapshot_file = "disk_snapshot.txt"
with open(snapshot_file, "w") as file:
    file.write("document.txt\n")
    file.write("image.jpg\n")
    file.write("report.pdf\n")

print("\nDisk snapshot saved.")

# -------------------------------
# Dead Acquisition (System OFF)
# -------------------------------
print("\nDEAD ACQUISITION")
print("-------------------------")

print("Reading saved disk snapshot:")

with open(snapshot_file, "r") as file:
    print(file.read())

print("Running Processes: Not Available")
print("RAM Data: Cannot be recovered after shutdown.")
```
