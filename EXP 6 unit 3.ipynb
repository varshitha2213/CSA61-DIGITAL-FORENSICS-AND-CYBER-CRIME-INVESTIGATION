Experiment 6: Bit Stream Copy Verification.

```python
import os
import shutil
import hashlib

# -------------------------------
# Create a Simulated Storage Device
# -------------------------------
original_file = "storage_device.bin"

with open(original_file, "wb") as f:
    f.write(b"This is the original storage device data.")

# -------------------------------
# Create Bit Stream Copy
# -------------------------------
copy_file = "bitstream_copy.bin"

shutil.copyfile(original_file, copy_file)

print("BIT STREAM COPY CREATED")
print("-" * 40)

# -------------------------------
# Compare File Sizes
# -------------------------------
original_size = os.path.getsize(original_file)
copy_size = os.path.getsize(copy_file)

print("Original Size :", original_size, "bytes")
print("Copy Size     :", copy_size, "bytes")

# -------------------------------
# Calculate SHA-256 Hash
# -------------------------------
def sha256_hash(filename):
    hash_obj = hashlib.sha256()
    with open(filename, "rb") as f:
        while True:
            data = f.read(4096)
            if not data:
                break
            hash_obj.update(data)
    return hash_obj.hexdigest()

original_hash = sha256_hash(original_file)
copy_hash = sha256_hash(copy_file)

print("\nOriginal SHA-256 :", original_hash)
print("Copy SHA-256     :", copy_hash)

# -------------------------------
# Verification
# -------------------------------
if original_size == copy_size and original_hash == copy_hash:
    print("\nVerification Successful!")
    print("The bit stream copy is a forensically exact replica.")
else:
    print("\nVerification Failed!")
    print("The copy is not identical to the original.")
```
