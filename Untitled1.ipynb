```python
patterns = {
    b"\xFF\xD8\xFF": "JPEG Image",
    b"%PDF": "PDF Document",
    b"PK\x03\x04": "ZIP Archive"
}

# Create a sample storage file
with open("storage.bin", "wb") as f:
    f.write(b"RandomData")
    f.write(b"%PDF")
    f.write(b"Sample PDF Content")
    f.write(b"More Random Data")

with open("storage.bin", "rb") as f:
    data = f.read()

found = False

for pattern, filetype in patterns.items():
    position = data.find(pattern)
    if position != -1:
        print(filetype, "found at byte", position)
        found = True

if not found:
    print("No recoverable file signatures found.")
```
