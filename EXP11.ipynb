```python
from collections import defaultdict

traffic_logs = [
    ("192.168.1.5", 22),
    ("192.168.1.5", 23),
    ("192.168.1.5", 80),
    ("192.168.1.5", 443),
    ("192.168.1.5", 21),
    ("192.168.1.10", 80),
    ("192.168.1.10", 443),
    ("192.168.1.20", 22),
]

ports = defaultdict(set)

for ip, port in traffic_logs:
    ports[ip].add(port)

print("Port Scan Detection")
print("-" * 30)

threshold = 4

for ip, scanned_ports in ports.items():
    if len(scanned_ports) > threshold:
        print(f"Possible Port Scan from {ip}")
        print("Ports:", sorted(scanned_ports))
```
